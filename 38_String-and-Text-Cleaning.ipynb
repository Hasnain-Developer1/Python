```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Category': ['Electronics', 'Furtinure', 'Home-Appliances', 'electronics', 'FURNITURE']
})

print(df)
```

```python
# strip spaces and convert to lowercase
df['Category_clean'] = df['Category'].str.strip().str.lower()
print(df)
```

```python
# Replace hyphen with spaces
df['Category_clean'] = df['Category_clean'].str.replace('-', " ")
print(df)
```
