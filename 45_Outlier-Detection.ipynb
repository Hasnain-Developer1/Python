# Outlier

outliers are the data points that fall out side the normal range, representing extreme values that can distort the mean and median of the dataset

# Methods of Outlier Identifyication and Removal

- IQR(Inter Quantile range) Method
- z-score Method
- Domain-Driven threshold

```python
# Load DataSet
import pandas as pd

df = pd.read_csv(
    "ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
```

```python
# IQR Method
Q1 = df["revenue"].quantile(0.25)
Q3 = df["revenue"].quantile(0.75)
IQR = Q3 - Q1

lower_bond = Q1 - 1.5 * IQR
upper_bond = Q3 + 1.5 * IQR

outliers_iqr = df[(df["revenue"] < lower_bond) | (df["revenue"] > upper_bond)]

# IQR(Inter Quantile Range): Measure the middle 50% of the data. robust the skewed distribution (like revenue)
```

```python
outliers_iqr.shape  # verification
```

```python
outliers_iqr[["revenue", "product_category"]].head()
```

```python
df = df[(df["revenue"] >= lower_bond) & (df["revenue"] <= upper_bond)]
df.shape
```

```python
# z-score Method 
from scipy.stats import zscore

df = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
df["revenue_z"] = zscore(df["revenue"])
outlier_z = df[df["revenue_z"].abs() > 3]

# z score shows how many standard deviation value is form the mean
# Rule: |Z|>3 -> Outlier
```

```python
outlier_z.shape  # verification
```

```python
df = df[df["revenue_z"].abs() <= 3]
df.shape
```
