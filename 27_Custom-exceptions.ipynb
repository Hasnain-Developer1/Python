```python
class InvalidAgeError(Exception): # when we will create custom exception we will inherit it from exception class
    pass
```

```python
age = int(input("Enter your age"))
if age < 0:
    raise InvalidAgeError(f"Age cannot be negative, {age}")
```

```python
try:
    age = int(input("Enter age: "))
    if age < 0:
        raise InvalidAgeError(f"Age cannot be negative, {age}")
except InvalidAgeError as e:
    print(e)
```
