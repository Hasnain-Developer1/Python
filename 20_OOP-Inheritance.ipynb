# Inheritance

Inheritance allows a new class to reuse the properties & methods of and existing class.

```python
class Animal:
    def speak(self):
        print("some sound")
```

```python
class Dog(Animal):
    pass  # pass keyword ka mtlb hota ha abhe ham is ma koi code ni likhna baad ma likha ga abhe chala do
```

```python
Dog1= Dog()
```

```python
Dog1.speak()
```

```python
class Dog(Animal): # class Dog is inheriting the class Animal
    def bark(self):
        print("Woof!")
```

```python
childing = Dog()
```

```python
childing.bark()
```

```python
childing.speak()
```
