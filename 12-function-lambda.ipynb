# Lambda Function

### A lambda Function in Python is a smaal anonymous function written in a single line

```python
add = lambda a,b: a+b
```

```python
print(add(3,5))
```

```python
double_it = lambda n: n*2
```

```python
print(double_it(5))
```

```python
nums = [1, 2, 3, 4]
squares = list(map(lambda x : x*x, nums))  # Yahan nums isliye likha hai kyun ke map() ko batana hota hai kis list par function chalana hai.
```

```python
print(squares)
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
odd_numbers = list(filter(lambda x : x % 2 != 0, numbers))
```

```python
print(odd_numbers)
```
