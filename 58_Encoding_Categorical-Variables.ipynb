```python
import pandas as pd
```

```python
# Sample Data
df = pd.DataFrame({
    "city": ["Lahore", "Karachi", "Islamabad", "Lahore", "Karachi"],
    "segment": ["A", "B", "A", "C", "B"],
    "sale": ["200", "240", "180", "300", "220"]
})

print(df)
```

```python
# one-hot encode
encoded = pd.get_dummies(df, columns=["city", "segment"], drop_first=True)
print(encoded)
```

```python
from sklearn.preprocessing import OrdinalEncoder

df2 = pd.DataFrame({
    "size": ["Small", "Mediam", "Large", "Mediam", "Large"]
})

print(df2)
```

```python
encoder = OrdinalEncoder(categories=[["Small", "Mediam", "Large"]])
df2["size_encoder"] = encoder.fit_transform(df2[["size"]])
print(df2)
```
