```python
import pandas as pd
```

```python
data = {
    'Name': ['Ali', 'Ahmad', 'Hasnain', 'Ayyan'],
    'Age':  [20, 21, 20, 19],
    'City': ['Layyah', 'BWP', 'Multan', 'RYK'],
    'Salary': [100000, 120000, 110000, 90000]
}
df = pd.DataFrame(data)

print("Original DataFrame: \n", df)
```

```python
#1. Label Based selection (.loc)
subset_loc = df.loc[1:3, ['Name', 'Salary']]
print("Subset with loc \n ", subset_loc)
```

```python
#1. Integrated Based selection (.loc)
subset_iloc = df.iloc[0:2, 1:3] # last index will not be included in iloc
print("Subset with iloc \n ", subset_iloc)
```

```python
# 3. boolean Filtering
high_salary = df[df['Salary'] > 100000] # this is condition, it will print those rows whose salary is grater then 100000
print("High salary filter: \n ", high_salary)
```

```python
# 4. Safe assignment using .loc
df.loc[df['Age'] < 30, 'Salary'] > 65000  # Salary fixed to all those whose age is less then 30
print("low age salary change: \n ", df.loc[df['Age'] < 30, 'Salary']) # print the salary of all those whose age is less then 30
```
