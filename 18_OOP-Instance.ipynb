```python
class student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    #Self keyword tells that these and instance variables
    
    def display_info(self):
        print(f"Student name : {self.name}")
        print(f"Student age : {self.age}")
        print(f"Student Grade : {self.grade}")

    def is_eligible(self):
        if self.age >= 15:
            print(f"{self.name} is eligible for admission")
        else:
            print(f"{self.name} is not eligible for admission")
```

```python
# Creating object
student1 = student("Ali", 20, "10th")
student2 = student("Ahmad", 19, "11th")
```

```python
# Accessing variable
print(student1.name)
print(student2.grade)
```

```python
student1.display_info()
student2.is_eligible()
```

```python
student2.display_info()
student2.is_eligible()
```
