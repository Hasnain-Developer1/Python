```python
class Dog:
    def __init__(self):
        self.name = "buddy" #Default methos that assign the default value
    def bark(self):
        print(f"{self.name} says woof!")
```

```python
dog1 = Dog()
dog1.bark()
```

```python
class car:
    def __init__ (self, color, brand):
        self.color = color
        self.brand = brand
```

```python
car1 = car("green", "toyota")
print(car1.color)
print(car1.brand)
```

```python
car1.color = "blue"
car2 = car("White", "Hinda")
print(car2.color)
print(car2.brand)
print(car1.color) # car1 color changed
```
