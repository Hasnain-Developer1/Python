```python
import numpy as np
```

```python
arr = np.array([1,2,3])
print(arr)
```

```python
print(np.zeros((2,3))) # it will create a matrix of two rows and three columns all values zero
```

```python
print(np.ones((2,3))) # it will create a matrix of two rows and three columns all values one
```

```python
print(np.full((2,3), 7)) # 2 rows and 3 columns all values will be 7
```

```python
print(np.arange(0, 10, 2)) # print from 0 to 10  and step size is 2, 0 will be included and 10 will not be included
```

```python
print(np.linspace(0, 1, 5)) # it mean print from 0 to 1 and give me 5 parts
```

```python
print(np.eye(3)) # Identity matrix of 3 rows and 3 columns
```

```python
print(np.diag([1,2,3])) # Diagnal values will be 1, 2, 3 and non-diagonal values will be 0
```

```python
np.array([1,2,3], dtype=float) # datatyep will be float see result is 1., 2., 3.
```

```python
arr = np.array([1.5, 2.7, 3.9])
print(arr.astype(int)) # it mean print as integer
```
