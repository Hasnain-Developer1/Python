```python
import numpy as np
import time
```

```python
arr = np.array([[1,2,3], [4,5,6]])
print(arr.shape) # (2,3) Two rows Three columns
print(arr.ndim) # 2
print(arr.dtype) # int64 (or platform dependent) mean it is a 64bit integer
```

```python
py_list = list(range(1_000_000))
py_result = []
start_time = time.time()
for i in py_list:
    py_result.append(i+5)
end_time = time.time()
print("Numpy time: " , end_time - start_time)
```

```python
np_array = np.arange(1_000_000)
start_time = time.time()
np_result = np_array + 5
end_time = time.time()
print("Numpy Array time: " , end_time - start_time)
```
