# Polymorphism

polymorphism refers to the ability of a function or method to exist in multiple forms.

```python
class Dog:
    def speak(self):
        print("Woof!")
```

```python
class Cat:
    def speak(self):
        print("Meow!")
        
```

```python
for pet in [Dog(), Cat()]:
    pet.speak()
```

```python
class vehicle:
    def move(self):
        print("Vehicle is moving")
```

```python
class car(vehicle):
    def move(self):
        print("car is driving")
```

```python
car1 = car()
```

```python
car1.move()
```

```python
class calculator:
    def add(self, a, b=0, c=0):
        return a + b + c
```

```python
cal1 = calculator()
```

```python
print(cal1.add(5))
```

```python
print(cal1.add(5, 10))
```

```python
print(cal1.add(5, 10, 15))
```
