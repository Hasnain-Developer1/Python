#### Seaborn integrates smoothly with Matplotlib for customization.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("healthcare_patient_analytics_seaborn.csv")
```

```python
df = df.sample(500)
# Axes Level example scatterplot
sns.scatterplot(
    data = df,
    x = "treatment_cost",
    y = "recovery_score",
    hue = "visit_type"
)

plt.title("Treatment COST vs Recovery Score")
plt.show()
```

```python
# Figure Level Example Relplot
sns.relplot(
    data = df,
    x = "treatment_cost",
    y = "recovery_score",
    hue = "visit_type",
    col = "gender"
)

plt.title("Treatment COST vs Recovery Score")
plt.show()
```

```python
# Applying seaborn Theme
sns.set_theme(style="whitegrid")
```

```python
# recreating a Matplotlin in Seaborn 

# matplotlib

plt.scatter(df["treatment_cost"], df["recovery_score"])
plt.xlabel("Treatment Cost")
plt.ylabel("Recovery Score")
```
