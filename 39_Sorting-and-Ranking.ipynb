```python
import pandas as pd
```

```python
df = pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],
    dtype={
        "Category": "Category",
        "Region": "Category"
    }
)
df.head()
```

```python
# Sort by Age ascending, then salary descending
df_sorted = df.sort_values(by=['Quantity', 'Sales'], ascending=[False, True])
print("Sorted the DataFrame")
df_sorted.head()
```

```python
# rank salary with Age Group
df['Sales_rank'] = df.groupby('Quantity')['Sales'].rank(method='dense', ascending=False)
print("\nSales rank within quantity Group:\n", df)
```
