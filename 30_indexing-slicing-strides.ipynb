```python
import numpy as np
```

```python
arr = np.array([[10,20,30],
               [40,50,60],
               [70,80,90]])
```

```python
print(arr[0,1]) # 20 (row 0 and column 1)
print(arr[2,2]) # 90 (row 2 and column 2)
```

```python
print(arr[-1, -2]) # 80 (last row and 2nd last column)
print(arr[-3, -3]) # 10 (first row and first column)
```

```python
# General form arr[start:stop:step]
print(arr[0:2, 1:3]) # rows 0-1, columns 1-2
```

```python
print(arr[::2, ::2]) # every 2nd row and 2nd columns 2 rows ka jump lya aur 2 columns ka jump lya
```

```python
slice_arr = arr[0, :2] # view(not a copy) 
slice_arr[0] = 999
```

```python
print(arr[0]) # original changes
```

```python
copy_arr = arr[1].copy() # independent copy
copy_arr[0] = 555
```

```python
print(arr[1]) # original unchnaged
```
