```python
num = int(input("Enter a number: "))
print(10 / num)
```

```python
try:
    num = int(input("Enter a number: "))
    print(10 / num)
except ZeroDivisionError:
    print("You cannot divide by zero")
except ValueError:
    print("that's not a number")
except Exception as e:
    print(f"an unexpected error occured {e}")
```

```python
try:
    print("Running code...")
except:
    print("Error Found")
else:
    print("No error Found!")
finally: # this finally keyword code must run even exception occur or not
    print("Done")
```
