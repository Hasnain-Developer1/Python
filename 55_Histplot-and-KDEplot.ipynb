```python
# Add the health Care Dataset
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("healthcare_patient_analytics_seaborn.csv")
```

```python
# histplot
# senerio - Treatment Cost Distribution

sns.histplot(
    data = df,
    x = "treatment_cost",
    bins = 30
)

plt.title("Distribution of Treatment Cost")
plt.show()
```

```python
# Compairing Distribution across Categories
# Scenario - Routine vs Emergency Visits

sns.histplot(
    data = df,
    x = "readmission_risk",
    hue = "visit_type",
    multiple = "stack",
    bins = 30
)

plt.title("readmission by Visit Type")
plt.show()
```

```python
# kdeplot - provide a smooth estimate of the distribution
sns.kdeplot(
    data = df,
    x = "recovery_score",
    hue = "visit_type",
    fill = True
)

plt.title("KDE of Recovery Score by visit Type")
plt.show()
```

```python
# ECDF - Shows Comulative proportions
sns.ecdfplot(
    data = df,
    x = "length_of_stay_days",
    hue = "visit_type",
)

plt.title("ECDF of length of Stay Days")
plt.show()
```
