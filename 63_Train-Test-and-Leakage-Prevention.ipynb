```python
# Load Dataset
import pandas as pd
import numpy as np

df = pd.read_csv("telecom_customer_churn_feature_engineering.csv")
```

```python
# len of df
len(df)
```

```python
# we will split it into 960 rows (80%) | 240 testing rows (20%)
```

```python
# Random split using Pandas
df_shuffled = df.sample(frac=1, random_state=42)

train_size = int(0.8 * len(df_shuffled))
train_df = df_shuffled.iloc[:train_size]
test_df = df_shuffled.iloc[train_size:]

len(train_df), len(test_df)
```

```python
# verify No Overlap
set(train_df.index).intersection(set(test_df.index))
```

```python
# Leakage Example - wrong Practice
# Wrong Scalling - Before Split
income_mean = df["monthly_income"].mean()
income_std = df["monthly_income"].std()

df["income_scaled_wrong"] = (
    df["monthly_income"] - income_mean
) / income_std

df[["monthly_income", "income_scaled_wrong"]].head()
```

```python
# What went wrong?
# test Data influenced mean & std
# Model indirectly "saw the Future"
# Correct approach: 1- Split Data first 2- Compute mean/std on train only 3- Apply some values to test
```

```python
# Separate Feature and Target
x_train = train_df.drop("churn", axis=1)
y_train = train_df["churn"]

x_test = test_df.drop("churn", axis=1)
y_test = test_df["churn"]

x_train.shape, x_test.shape
```
