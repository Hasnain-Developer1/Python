# Corelation and Covariance

#### Measure the relationship between two numberical values

## positive Corelation:

Both variables increase or decrease

## Negative Corelation:

One value increase other decrease

```python
import pandas as pd

df = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
```

```python
numeric_df = df[["quantity", "unit_price", "discount", "delivery_days", "revenue"]]
# covariance matrix
print("\ncovariance:\n", numeric_df.cov())
```

```python
# corelation matrix
print("\ncorealtion:\n", numeric_df.corr())
```

```python
numeric_df.corr()["revenue"].sort_values(ascending=False)  # Focus on revenue Relationship
```

```python
numeric_df.corr().values.min(), numeric_df.corr().values.max()  # Verification
# Expevted minimum >= -1 and maximum <= +1
```
