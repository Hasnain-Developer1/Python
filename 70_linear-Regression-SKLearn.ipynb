```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[1], [3], [4], [5], [7]]) 
Y = np.array([3, 5, 7, 9, 11])
```

```python
# Plot 1: Scatter plot before regression
plt.figure(figsize=(6, 4))
plt.scatter(X, Y, color='blue')
plt.title("Scatter Plot Before Linear Regression")
plt.xlabel("X values")
plt.ylabel("Y values")
#plt.grid(True) 
plt.show()
```

```python
model = LinearRegression() # Create a Linear Model Regression
model.fit(X, Y) # Fit the model to the data
y_pred = model.predict(X) # Predict using the fitted model

print("Predicted Values:", y_pred) # print the predicted values
```

```python
model = LinearRegression() # Create a Linear Model Regression
model.fit(X, Y) # Fit the model to the data
y_pred = model.predict(X) # Predict using the fitted model

print("Predicted Values:", y_pred) # print the predicted values
```

```python

# Plot 2: Scatter + Best Fit Line
plt.figure(figsize=(6, 4))
plt.scatter (X, Y, color='blue', label='Original Data') 
plt.plot(X, y_pred, color='red', label='Best Fit Line') 
plt.title("Scatter Plot with Regression Best Fit Line") 
plt.xlabel("X values") 
plt.ylabel("Y values") 
plt.legend()
#plt.grid(True)
plt.show()
```

```python
m = model.coef_
b = model.intercept_
```

```python
print("Slope (m): ", m)
print("Y intercept (b): ", b)
```
