```python
from langchain_community.embeddings import HuggingFaceEmbeddings
# from langchain_huggingface import HuggingFaceEmbeddings
# 1. Initialize Embeddings ---
# We use a lightweight model suitable for CPU

embedding_function = HuggingFaceEmbeddings (model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
content = """
LangChain is a framework for developing applications powered by language models.
It enables applications that are context-aware and reason across problems. Key components of LangChain include Models, Prompts, Chains, and Memory. chains allow you to combine multiple components together.
RAG (Retrieval-Augmented Generation) is a popular technique used in LangChain.
"""
```
