# Class

A class is a blueprint for creating object. Define attributes (Data) & Methods (behavior).

# Attrubutes

data stored in the object (name, age)

# Methods

Actions that the object can perform(speak, run)

# Benefits of Object Oriented Programming

- Simplifies complex problems
- makes teamwork and collaboration easier
- Encourage reuse and scalability
- forms the foundation of modern programming languages(python, JAVA, C#, etc.)
