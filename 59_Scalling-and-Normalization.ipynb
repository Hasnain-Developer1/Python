```python
import pandas as pd
import numpy as np
df = pd.read_csv("telecom_customer_churn_feature_engineering.csv")
```

```python
# Inspect raw values
df[["monthly_income", "monthly_bill"]].head() # Income and Bill values are on very Different sale
```

```python
# Why Scalling is required
df[["monthly_income", "monthly_bill"]].describe() # Income dominate numerically - Models will focus more on income
```

```python
# z-score scalling standardization (Manual)
# 1. Calculate mean & std
income_mean = df["monthly_income"].mean()
income_std = df["monthly_income"].std()

print("Mean:", income_mean)
print("std:", income_std)
```

```python
# 2. Apply z score Formula
df["income_zscore"] = (
    df["monthly_income"] - income_mean
) / income_std

df[["monthly_income", "income_zscore"]].head()
```

```python
# Verify zscore scalling
df["income_zscore"].mean(), df["income_zscore"].std()
```

```python
# min-max scalling normalization (manual)
bill_min = df["monthly_bill"].min()
bill_max = df["monthly_bill"].max()

print("Min:", bill_min)
print("Max:", bill_max)
```

```python
# Apply min-max formula
df["bill_minmax"] = (
    df["monthly_bill"] - bill_min
) / (bill_max - bill_min)

df[["monthly_bill", "bill_minmax"]].head()
```

```python
# verify minmax scalling
df["bill_minmax"].min(), df["bill_minmax"].max()
```

```python
# show minimum maximum and mid value together
df.loc[
    [df["monthly_bill"].idxmin(),
     df["monthly_bill"].idxmax(),
     df["monthly_bill"].sort_values().index[len(df)//2]],
    ["monthly_bill", "bill_minmax"]
]
```

```python
# Side by Side Comaprison
df[["monthly_income", "income_zscore",
   "monthly_bill", "bill_minmax"]].head()
```
