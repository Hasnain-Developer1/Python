# What is Dictionay

##### Stores data as key value pairs

```python
student = {
    "name": "Hasnain Malik",
    "age": 20,
    "course": "Artificial Intelligence"
}

print(student)
```

```python
print(student["name"])
print(student.get("age"))
```

```python
student["age"] = 20 # Update
student["city"] = "Layyah" # add
student.pop("course") # remove the course
```

```python
print(student)
```

# Looping Dictinary

```python
for key, value in student.items():
 print(key, ":", value)
```
