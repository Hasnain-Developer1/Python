# Specila Function

A special function defines built-in behavior  automatically associated with certain operations.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
    def __str__(self):
        return f"'{self.title}' by {self.author}"
```

```python
b = Book("1980", "George orwell")
```

```python
print(b)
```

```python
class playlist:
    def __init__(self, songs):
        self.songs = songs
    def __len__(self):
        return len(self.songs)
    def __add__(self, other):
        return playlist(self.songs + other.songs)
            
```

```python
p1 = playlist(["Song1", "song2"])
```

```python
p2 = playlist(["song3"])
```

```python
print(len(p1)) # it will print 2 as there are two songs in p1
```

```python
print(len(p1 + p2)) # it will print 3 as there are three songs in p2
```

```python
class point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __str__(self):
        return f"({self.x}, {self.y})"
    def __add__(self, other):
        if not isinstance(other, point):
            return NotImplemented
        return point(self.x + other.x, self.y + other.y)
```

```python
p1 = point(2, 3)
```

```python
p2 = point(4, 1)
```

```python
print(p1 + p2)
```
