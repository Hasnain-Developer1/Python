Encapsulation means keeping all the data (attrubutes) & methods of an object inside that object. 
mean Hiding data

# Private Attributes

Private attributes are those attributes that can only be used within class's own methods.

```python
class student:
    def __init__(self, name, grade):
        self.__grade = grade  # These two Underscores"__" tells this is private attribute
student1 = student("Ali", 90)
print(student1.__grade) # Error
```

```python
class student:
    def __init__(self, name, grade):
        self.__grade = grade 
        
    def get_grade(self):
        return self.__grade
        
student = student("Ali", 90)
print(student.get_grade()) 
        
```

```python
class student:
    @property
    def grade(self):
        return self.__grade
    def __init__(self, name, grade):
        self.__grade = grade
    def get_grade(self):
        return self.__grade
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Invalid Grade")

student = student("Ali", 20)
student.set_grade(25)
student.get_grade()
```

# Getter and Setter

getters are functions used to retrieve the data attributesof a class.


Setter are funcvtions used to assign or modify the data attributes of a class

```python
class student:
    @property
    def grade(self):
        return self.__grade
    def __init__(self, name, grade):
        self.__grade = grade
    def get_grade(self):
        return self.__grade
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Invalid Grade")

student = student("Ali", 20)
student.set_grade(-25)
print(student.get_grade())
```
