```python
with open("demo.txt", "r") as file:
    data = file.read() # write in the word "with" indentation
    print(data)

#File Close automatically here
```

## Using Context Managers(Other Statements)


- open(): The standard file context manager for
reading/writing files.
- tempfile.TemporaryFile() / Named TemporaryFile(): Creates temporary files that delete themselves
automatically.
- zipfile.ZipFile(): Opens .zip files for reading/writing.
- contextlib.ExitStack: Lets you manage multiple files at
once easily.
- pathlib.Path().open()
Modern, object-oriented file opening via pathlib.

# Exception handling

An exception is an error that occur during the execution of the program

#### We use excepotion so that if the file does not exist or path is incorrect, the program should not crash and will run just a error "File not Found"

```python
with open("missing.txt", "r") as file:
    print(file.read())
```

```python
try:
    with open("missing.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File Not Found")
```

```python
filename = "my_info.txt"
name = "GitHub Copilot"
age = 1
```

```python
try:
    with open(filename, "r") as fin:
        print(fin.read())
except FileNotFoundError:
    with open(filename, "w") as fout:
        fout.write(f"{name}\n{age}")
    with open(filename, "r") as fin:
        print(fin.read())


        
    
```
