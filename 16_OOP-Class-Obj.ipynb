```python
car1 = car()
car1.drive()
```

```python
class car:
    color = "red"
    def drive(self):
        print(f"{self.color} car is moving")
```

```python
car1 = car()
car1.drive()
```

```python
class car:
    color = "red"
    def drive(self):
        print(f"{self.color} car is moving")
    def setColor(self, new_color):
        self.color = new_color
        
```

```python
car2 = car()
car2.setColor("green")
car2.drive()
```
