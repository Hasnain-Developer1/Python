# For loop

```python
for i in range(3):
    print("Python is fun")
```

```python
for char in "Python":
    print(char)
```

```python
i = 4
rows = i
cols = i
matrix = []
for r in range(rows):
    row = []
    for c in range(cols):
        row.append(0)
    matrix.append(row)
for row in matrix:
    print(' '.join(map(str, row)))
```

```python
for level in range(1, rows + 1): # row + 1 is written because last number in nor included in range if we dont add 1 it mean it will run up to 3
    print(' '* (rows - level) + '*' * (2 * level - 1)) 
```

# While Loop

```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

# Nested loop

```python
for i in range(3):
    print("Outer Loop")
    for j in range(3):
        print("   Inner Loop")
```
