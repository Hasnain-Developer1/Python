```python
import numpy as np
```

```python
# simulate 100k coins flips
rng = np.random.default_rng(seed=43)
flips = rng.integers(0, 2, size=100000) # 0 is tail and 1 = head
prob_heads = np.mean(flips)
print("Estimated probability of head: ", prob_heads)
```

```python
import time

#python loop
start = time.time()
squares = [i**2 for i in range(1000000)]  # Loop is slow
end = time.time()
print("Loop time: ", end - start)

#Numpy vactorized
start = time.time()
arr = np.arange(1000000) #numpy is fast
squares_np = arr**2
end = time.time()
print("Vectorized Time: ", end - start)
```
