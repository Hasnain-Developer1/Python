```python
import pandas as pd

df = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
```

```python
# Verify

df.info()
```

```python
# map() = label high value orders
df["payment_type"] = df["payment_method"].map({  # this method will apply on payment_method column and create a new column "payment_tyoe"
    "Card": "Digital",
    "Wallet": "Digital",
    "COD": "Cash"
})

# Explanation: each value is mapped individually fast and simple
```

```python
# verify
df["payment_type"].unique()  # It will show that there are new unique value 'Digital' and 'cash'
```

```python
df[["payment_method", "payment_type"]].head()
```

```python
df.groupby("payment_type").size()
```

```python
# apply Create revenue category

def revenue_category(row):
    if row["revenue"] > 500:
        return "High"
    elif row["revenue"] >= 200:
        return "Mediam"
    else:
        return "Low"

df["revenue_category"] = df.apply(revenue_category, axis=1) # new revenue_category column will be created based on vevenue using the above if else formula


# why apply? Decision depends on multiple columns.
```

```python
# verify
df["revenue_category"].value_counts() # check categories
```

```python
df[["revenue", "revenue_category"]].sample(10)
# Revenue > 500 -> High
# Revenue 200 - 500 -> Mediam
# < 200 -> Low
```

```python
# pipe() - Method Chaining 
def add_profit(data):
    data["profit"] = data["revenue"] * 0.20
    return data

def add_net_revenue(data):
    data["net_revenue"] = data["revenue"] - data["profit"]
    return data

      # series of operations to be performed on dataset

df = (
    df
    .pipe(add_profit)
    .pipe(add_net_revenue)
)
```
