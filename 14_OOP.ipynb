# OOP

### Encapsulation

- in programming, related data in encapsulated in a single object.
- The functions related to those variables are also grouped together

### Object

- combination of related variables and functions, Stores Data(attributes), perform actions(methods)

# Data Abstraction

- Abstraction means that we do not need to show all our data and functions to the outside world

### Benefit of Data Abstraction

- permotes code reuseability
- Enhance modularity
- improves maintainability
- Makes program closer to real world thinking
