# What is EDA?

 EDA Stands for Exploratory Data Analysis is the process of understanding the Data before modelling.

# Key Goals of EDA

- Understanding Patterns, trends and anomalies
- Check Data Quality (Missing values, Duplicate Values)
- Understand Feature Types

# EDA Mindset and CheckList

- Be curious why and how
- look for relationship between features
- Document Assumptions and Findings

# Functions of EDA Report

- (.shape) to see dimensions of Data
- (.dtype) tp see the columns type
- (.isna().sum()) See the Missing value
- (.describe) Summary Statistics
- (.value_counts()) get unique Values and categories

# Ket Takeaways

- Always Define goals before starting EDA
- Idenrtify Target and feature Type early
- use EDA Questions to guide analysis
- Document Data Quality issues and Assumptions
