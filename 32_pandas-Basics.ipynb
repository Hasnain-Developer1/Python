```python
import pandas as pd
```

```python
# create a pandas series
s = pd.Series([10, 20, 20, 30, 40], index=['a', 'b', 'c', 'd', 'e']) # S should be capital in series
print(s)
```

```python
# index as values
print("Values: ", s.values)
print("Index: ", s.index)
print("Dtype: ", s.dtype)
```

```python
print(s.head)
```

```python
print(s.tail)
```

```python
print(s.value_counts()) # it tells a value is occured how many times 
```
