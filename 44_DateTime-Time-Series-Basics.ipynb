# DateTime and Time Series Basics

```python
# Load DataSet
import pandas as pd

df = pd.read_csv(
    "ecommerce_sales_analytics_5000.csv", parse_dates=["order_date"])
```

```python
# Setting DateTime Index
df = df.set_index("order_date")
```

```python
df.index
```

```python
# Monthly revenue Analysi
monthly_revenue = df["revenue"].resample("ME").sum()  # ME(Month End) it was 'M' only but Now in New versions of Python It is 'ME'
monthly_revenue
```

```python
# Trend Interpretation
monthly_revenue.pct_change() # percentage
```

```python
monthly_revenue.index.is_monotonic_increasing # Verify
```

```python
print(monthly_revenue.sum(), df["revenue"].sum())  # check
monthly_revenue.sum() <= df["revenue"].sum() # check
```
